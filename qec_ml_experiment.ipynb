# Machine Learning for Quantum Error Correction

Small-scale experiment comparing a rule-based syndrome decoder with a logistic-regression decoder for a three-qubit bit-flip code.

```python
from src.evaluation import evaluate_decoders

results = evaluate_decoders()
results
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.plot(results['noise_probability'] * 100, results['baseline_accuracy'], marker='o', label='Baseline decoder')
plt.plot(results['noise_probability'] * 100, results['ml_accuracy'], marker='o', label='ML decoder')
plt.xlabel('Physical bit-flip probability (%)')
plt.ylabel('Decoder accuracy')
plt.title('QEC Decoder Performance')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
